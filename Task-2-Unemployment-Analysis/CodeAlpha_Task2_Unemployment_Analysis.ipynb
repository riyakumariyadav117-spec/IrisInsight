```python

# Import necessary libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

df = pd.read_csv("Unemployment in India.csv")

print(df.head())
```

```python

# Understanding our dataset

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python

# STEP 1: Dataset Investigation

# Remove accidental spaces from column names
df.columns = df.columns.str.strip()

# 1. Dataset dimensions
print("1. DATASET SIZE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 2. Column information
print("\n2. COLUMN INFORMATION")
print(df.dtypes)

# 3. Missing values
print("\n3. MISSING VALUES")
print(df.isnull().sum())

# 4. Duplicate records
print("\n4. DUPLICATE RECORDS")
print("Duplicates:", df.duplicated().sum())

# 5. Statistical summary
print("\n5. STATISTICAL SUMMARY")
display(df.describe(include="all").T)

# 6. Sample records
print("\n6. SAMPLE DATA")
display(df.head(10))
```

```python
# Create a backup of the original dataset
df_original = df.copy()

print("Original dataset backup created successfully!")
```

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
print("Completely empty rows:", df.isnull().all(axis=1).sum())
```

```python
# Remove rows where all columns are empty
df.dropna(how='all', inplace=True)

print("Missing rows removed successfully!")
print("New dataset shape:", df.shape)
```

```python
# Check duplicate records
print("Duplicates before cleaning:", df.duplicated().sum())

# Remove duplicate records
df.drop_duplicates(inplace=True)

print("Duplicates removed successfully!")
print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
# Convert Date column into datetime format
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

print("Date column converted successfully!")

print(df['Date'].dtype)
```

```python
print("FINAL DATASET INFORMATION")

print("\nDataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Original shape:", df_original.shape)
print("Current shape:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
print("Starting Date:", df['Date'].min())
print("Ending Date:", df['Date'].max())
```

```python
average_unemployment = df['Estimated Unemployment Rate (%)'].mean()

print("Average Unemployment Rate:",
      round(average_unemployment, 2), "%")
```

```python
state_unemployment = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

print(state_unemployment.sort_values(ascending=False))
```

```python
top_10 = state_unemployment.sort_values(ascending=False).head(10)

print(top_10)
```

```python
lowest_10 = state_unemployment.sort_values().head(10)

print(lowest_10)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10.values,
    y=top_10.index,
    hue=top_10.index,
    palette='Reds_r',
    legend=False
)

plt.title('Top 10 Regions with Highest Unemployment Rate')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('Region')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=lowest_10.values,
    y=lowest_10.index,
    hue=lowest_10.index,
    palette='Greens_r',
    legend=False
)

plt.title('Top 10 Regions with Lowest Unemployment Rate')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('Region')
plt.tight_layout()
plt.show()
```

```python
monthly_unemployment = df.groupby('Date')['Estimated Unemployment Rate (%)'].mean()

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker='o',
    linewidth=2
)

plt.title('Unemployment Rate Trend Over Time in India')
plt.xlabel('Date')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("Date Range:", df['Date'].min(), "to", df['Date'].max())

print("Overall Average Unemployment Rate:",
      round(df['Estimated Unemployment Rate (%)'].mean(), 2), "%")
```

```python
# Compare unemployment before and during the COVID period

df['Period'] = df['Date'].apply(
    lambda x: 'Before COVID' if x < pd.Timestamp('2020-03-01')
    else 'During COVID'
)

covid_comparison = df.groupby('Period')[
    'Estimated Unemployment Rate (%)'
].mean()

print(covid_comparison)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=covid_comparison.index,
    y=covid_comparison.values,
    hue=covid_comparison.index,
    palette='Set2',
    legend=False
)

plt.title('Unemployment Before and During COVID-19')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()
plt.show()
```

```python
area_unemployment = df.groupby('Area')[
    'Estimated Unemployment Rate (%)'
].mean()

print(area_unemployment)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=area_unemployment.index,
    y=area_unemployment.values,
    hue=area_unemployment.index,
    palette='Set2',
    legend=False
)

plt.title('Rural vs Urban Unemployment Rate')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()
plt.show()
```

```python
monthly_data = df.groupby('Date')[
    'Estimated Unemployment Rate (%)'
].mean()

highest_month = monthly_data.idxmax()
highest_rate = monthly_data.max()

print("Highest Unemployment Month:", highest_month)
print("Highest Unemployment Rate:", round(highest_rate, 2), "%")
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_data.index,
    monthly_data.values,
    marker='o',
    color='red'
)

plt.title('Monthly Unemployment Rate in India')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
area_unemployment = df.groupby('Area')[
    'Estimated Unemployment Rate (%)'
].mean()

print(area_unemployment)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=area_unemployment.index,
    y=area_unemployment.values,
    hue=area_unemployment.index,
    palette='Set2',
    legend=False
)

plt.title('Rural vs Urban Unemployment Rate')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()
plt.show()
```

# Conclusion – Unemployment Analysis with Python

This project analyzes unemployment trends in India using Python and data visualization techniques.

The dataset contains information about 28 regions of India, covering May 2019 to June 2020, with 740 records after data cleaning.

The analysis was performed using Pandas, NumPy, Matplotlib and Seaborn.

### Key Findings

1. The overall average unemployment rate was approximately **11.79%**.
2. Tripura recorded the highest average unemployment rate at approximately **28.35%**.
3. Meghalaya recorded the lowest average unemployment rate at approximately **4.80%**.
4. Odisha recorded an average unemployment rate of approximately **5.66%**.
5. The average rural unemployment rate was approximately **10.32%**, while the urban unemployment rate was approximately **13.17%**.
6. The analysis compared unemployment rates before and during the initial COVID-19 period.
7. Monthly trends were visualized to understand changes in unemployment over time.

### Final Conclusion

The analysis shows differences in unemployment rates across regions and between rural and urban areas. In this dataset, urban areas had a higher average unemployment rate than rural areas.
